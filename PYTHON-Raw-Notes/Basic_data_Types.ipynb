## $Data-Types$

```python
## Data Types example
# Integers, Floats, Strings, Booleans, Lists, Dictionaries, Tuples, Sets

# Integer
x = 10
print(x,type(x)) # <class 'int'>
```

```python
#Floats
y = 3.14
print(y,type(y)) # <class 'float'>
```

```python
#Strings
name = "Swastik"
print(name,type(name)) # <class 'str'>
```

```python
# Booleans
is_valid = True
print(is_valid,type(is_valid)) # <class 'bool'>
```

```python
# Lists
my_list = [1, 2, 3, "four", "five"]
print(my_list,type(my_list)) # <class 'list'>
```

```python
# Dictionaries
my_dict = {"name": "Swastik", "age": 30, "is_student": False}
print(my_dict,type(my_dict)) # <class 'dict'>
```

```python
# Truples
my_tuple = (1, 2, 3, "four", "five")
print(my_tuple,type(my_tuple)) # <class 'tuple'>
```

```python
# Sets
my_set = {1, 2, 3, "four", "five"}
print(my_set,type(my_set)) # <class 'set'>
```

```python
# ndarrays
my_array = np.array([1, 2, 3, 4, 5])
print(my_array,type(my_array)) # <class 'numpy.ndarray'>
```
