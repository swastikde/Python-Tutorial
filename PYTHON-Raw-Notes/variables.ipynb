```python
## Understanding variable types
a = 10
print(type(a))
b = 3.14
print(type(b))
c = 'Hello'
print(type(c))
d = True
print(type(d))
e = None
print(type(e))
f = [1, 2, 3]
print(type(f))
g = (1, 2, 3)
print(type(g))
h = {'name': 'Swastik', 'age': 30}
print(type(h)) 
i = {1, 2, 3}
print(type(i))
j = frozenset([1, 2, 3])
print(type(j))
```

```python
## Type Check & Conversion
a = 10
print(type(a))
b = float(a)
print(b)
print(type(b))
c = str(a)
print(c)
print(type(c))
d = int(b)
print(d)
print(type(d))
```

```python
## Dynamic Typing 
## Python allows you to change the type of a variable by assigning a new value of a different type to it. This is known as dynamic typing.
var = 10
print(var,type(var))

var = 'Hello'
print(var,type(var))

var = 3.14
print(var,type(var))
```
