# $Python-Control-Flow$

### $Conditional \ Statements$

```python
## If statement 
a = 10
b = 20
if a < b:
    print("a is less than b")
```

```python
## Else Statement 
a = 10
b = 20
if a < b:
    print("a is less than b")
else:
    print("a is greater than or equal to b")
```

```python
## Elif Statement
a = 10
b = 20
if a < b:
    print("a is less than b")
elif a == b:
    print("a is equal to b")
else:
    print("a is greater than b")
```

```python
## Nested conditional statements
a = 10
b = 20
if a < b:
    print("a is less than b")
    if a < 5:
        print("a is less than 5")
    else:
        print("a is greater than or equal to 5")
else:
    print("a is greater than or equal to b")
```

```python
## Even or odd number
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(f"{number} is an even number.")
else:
    print(f"{number} is an odd number.")
```

```python
## Check Prime Number
number = int(input("Enter a number: "))
if number > 1:
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            print(f"{number} is not a prime number.")
            break
    else:
        print(f"{number} is a prime number.")
```

```python
## Leap Year Check
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
## Simple calculator using if-else
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")
if operation == "+":
    result = num1 + num2
    print(f"The result of {num1} + {num2} is: {result}")
elif operation == "-":
    result = num1 - num2
    print(f"The result of {num1} - {num2} is: {result}")
elif operation == "*":
    result = num1 * num2
    print(f"The result of {num1} * {num2} is: {result}")
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
        print(f"The result of {num1} / {num2} is: {result}")
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid operation. Please enter one of +, -, *, /.")
```

# $Loops \ In \ Python$

#### $For \ Loop$

```python
## For Loop 
for i in range(5):
    print(f"Iteration {i}")
```

```python
## For Loop 
for i in range(2,5,1):
    print(f"Iteration {i}")
```

```python
## For Loop 
for i in range(2,10,2):
    print(f"Iteration {i}")
```

```python
## For Loop 
for i in range(10,1,-2):
    print(f"Iteration {i}")
```

```python
## Strings 
my_string = "Hello, World!"

for i in my_string:
    print(i)
```

#### $While \ Loop$

```python
count = 0

while count < 5:
    print(f"Count is: {count}")
    count += 1
```

```python
## Loop Control Statements: break, continue, pass

#Break statement example
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## star program using for loop
def star_program(n):
    for i in range(1, n + 1):
        print("*" * i)

star_program(5)
```

```python
## Star Program using while loop
def star_program(n):
    i = 1
    while i <= n:
        print("*" * i)
        i += 1
star_program(5)
```

```python
## Nested Loops Example
for i in range(1, 4):
    for j in range(1, 4):
        print(f"i: {i}, j: {j}")
```

```python
## Nested Loops Example
for i in range(1, 4):
    for j in range(1, 4):
        if i == j:
            break
        print(f"i: {i}, j: {j}")
```

```python
## Nested Loops Example
for i in range(1, 4):
    for j in range(1,4):
        if i == j:
            continue
        print(f"i: {i}, j: {j}")
```
