## $ Inheritence\ In\ Python $

## What is Inheritence in Python?
Inheritence is a fundamental concept in object-oriented programming (OOP) that allows a new class (called a child or subclass) to inherit attributes and behaviors (methods) from an existing class (called a parent or superclass). This promotes code reusability and establishes a natural hierarchical relationship between classes.

```python
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"Person will drive the {self.enginetype} Car")
car1 = Car(4,5,"Petrol")
car1.drive()
```

```python
class Tesla(Car): ## Inherit from Car class
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype)  ## inherit the attributes & methods from Car Class
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f"Tesla Supports Self Driving : {self.is_selfdriving}")

tesla1 = Tesla(4,5,"Electric",True)
tesla1.selfdriving()
tesla1.drive()
```

```python
## Multiple Inheritance
## Base clas 1
class Animal:
    def __init__(self,name):
        self.name = name 
    def speak(self):
        print("Subclass Must Implement this method")

### Base Class 2 
class Pet:
    def __init__(self,owner):
        self.owner = owner

### Derived Class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def speak(self):
        print(f"Name is {self.name} & Owner is {self.owner}")

case1 = Dog("Viky","Swastik")
case1.speak()
```

## $ Polymorphism\ in\ OOPS $

### What is polymorphism in OOPS?
Polymorphism is the ability of an object to take on many forms. It allows us to use a single interface to represent different types of objects. In Python, polymorphism is achieved through method overriding and duck typing.

##### Method Overriding

```python
## Base Class 
class Animal:
    def speak(self):
        return "Sound of the Animal"
## Derived Class 1       
class Dog(Animal):
    def speak(self):
        return "Bark Bark"
        
## Derived Class 2
class Cat(Animal):
    def speak(self):
        return "Meow Meow"

dog1 = Dog()
cat1 = Cat()    

print(dog1.speak())
print(cat1.speak())

```

```python
### Polymorphism with Functions & classes
## Base Class 
class Shape:
    def area(self):
        return "The Area of the Shape"
## Derived Class 1
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
## Derived Class 2
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

## A function  that demonstrates Polymorphism

def calculate_area(shape):
    print(f"The Area is : {shape.area()}")

rectangle1 = Rectangle(5,3)
circle1 = Circle(4)

calculate_area(rectangle1)
calculate_area(circle1)
```

## What is Polymorphism with Abstract Base Classes
Abstract Base Classes (ABCs) are a powerful tool in object-oriented programming that allow you to define abstract classes and methods. An abstract class is a class that cannot be instantiated and is meant to be subclassed. An abstract method is a method that is declared but contains no implementation. Subclasses of an abstract class are required to implement the abstract methods, ensuring a consistent interface across different implementations.    

```python
## Polymorphism with Abstract Base Classes
from abc import ABC, abstractmethod
## Define an Abstract Base Class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
## Derved Class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car Engine Started"

## Derived Class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle Engine Started"

### Function to demonstrate Polymorphism with Abstract Base Classes
def start_vehicle_engine(vehicle):
    print(vehicle.start_engine())   

## Create objects of the derived classes
car1 = Car()
motorcycle1 = Motorcycle()

start_vehicle_engine(car1)
start_vehicle_engine(motorcycle1)
```

## $ Encapsulation\ in\ Python $

## What is Encapsulation in Python
Encapsulation is one of the fundamental principles of Object-Oriented Programming (OOP) that refers to the bundling of data (attributes) and methods (functions) that operate on the data into a single unit, called a class. It also restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data. In Python, encapsulation can be achieved using access modifiers like public, protected, and private.

```python
## Encapsulation with Getter and Setter Methods
## Public, Protected and Private Access Modifiers in Python
class Person:
    def __init__(self,name,age):
        self.name = name  ## Public Variables
        self.age = age    ## Public Variables

person1 = Person("Swastik",30)
print(person1.name)
print(dir(person1))
print(get_name) ### Publicly Available Method
```

```python
class Person:
    def __init__(self,name,age):
        self.__name = name  ## Private Variable
        self.__age = age    ## Private Variable
person1 = Person("Swastik",30)
dir(person1)
```

```python
## Another Example
class Person:
    def __init__(self,name,age,gender):
        self.__name = name  ## Private Variable
        self.__age = age    ## Private Variable
        self.gender = gender  ## Public Variable

class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

employee1 = Employee("Swastik",30,"Male")
print(employee1.gender) ## This is a public variable and can be accessed directly
print(employee1.__name) ## This will give an error as __name is a private variable and cannot be accessed directly
```

```python
### Encapsulation with Getter and Setter Methods
class Person:
    def __init__(self,name,age):
        self.__name = name  ## Private Variable
        self.__age = age    ## Private Variable
    ## Getter Method for name
    def get_name(self):
        return self.__name

    ## Setter Method for name (change the value of private variable)
    def set_name(self,name):
        self.__name = name

person1 = Person("Swastik",30)
print(person1.get_name()) ## Accessing the private variable using getter method
```

```python
person1 = Person("Swastik",30)
print(person1.get_name())
person1.set_name("Swastik De")
print(person1.get_name()) ## Accessing the private variable using getter method after changing the value
```

## $ Abstarction\ in\ OOPS $

## What is Abstraction in Python?
Abstraction is the process of hiding the internal implementation details and showing only the functionality to the user. It helps in reducing complexity and allows the user to interact with the object at a higher level. In Python, we can achieve abstraction using Abstract Base Classes (ABC) from the abc module.

```python
from abc import ABC, abstractmethod
## abstract base class
class Veichle(ABC):
    def start_engine(self):
        print("Veicle is ready")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Veichle):
    def start_engine(self):
        return "Car Engine Started"
    
def operate_vehicle(vehicle):
    print(vehicle.start_engine())

car1 = Car()
operate_vehicle(car1)
```
