## $Operators$

```python
# Arithmetic Operators
a = 10
b = 3
print("Addition:", a + b)        # 13
```

```python
#Arithermetic Operators
a = 10
b = 3

addition = a + b
print("Addition:", addition)     # 13
subtraction = a - b
print("Subtraction:", subtraction)  # 7
multiplication = a * b
print("Multiplication:", multiplication)  # 30
division = a / b
print("Division:", division)     # 3.3333333333333335
floor_division = a // b
print("Floor Division:", floor_division)  # 3
modulus = a % b
print("Modulus:", modulus)       # 1
exponentiation = a ** b
print("Exponentiation:", exponentiation)  # 1000
```

```python
str1 = "Hello"
str2 = "World"
string_concatenation = str1 + " " + str2
print("String Concatenation:", string_concatenation)  # Hello World

print(str1 == str2)  # False
```

```python
a = 5
b = 10

print("a > b:", a > b)   # False
print("a < b:", a < b)   # True
print("a == b:", a == b) # False
print("a != b:", a != b) # True
```

```python
# Logical Operators
x = True
y = False
print("x and y:", x and y)  # False
print("x or y:", x or y)    # True
print("not x:", not x)      # False
print("not y:", not y)      # True
```
